# Notebook 03 — Exploratory Data Analysis


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")

BASE_DIR = Path.cwd()
if BASE_DIR.name == "notebooks":
    BASE_DIR = BASE_DIR.parent

PROCESSED_DIR = BASE_DIR / "data" / "processed"
df = pd.read_csv(PROCESSED_DIR / "IPL_Batting_Statistics_Cleaned.csv")
innings_df = pd.read_csv(PROCESSED_DIR / "IPL_Batting_Innings_Cleaned.csv")

display(df.describe())
```

```python
top_scorers = df.nlargest(
    10, "Runs"
)[["Player", "Season", "Matches_Played", "Runs", "Batting_Average", "Strike_Rate"]]
display(top_scorers.reset_index(drop=True).round(2))
```

```python
player_runs = (
    df.groupby("Player", as_index=False)["Runs"]
    .sum()
    .sort_values("Runs", ascending=False)
)
display(player_runs.head(10).round(2))
```

```python
season_summary = (
    df.groupby("Season", as_index=False)
    .agg(
        Total_Runs=("Runs", "sum"),
        Avg_Strike_Rate=("Strike_Rate", "mean"),
        Players=("Player", "nunique"),
    )
    .sort_values("Season")
)
display(season_summary.round(2))
```

```python
top_six_hitters = df.nlargest(10, "Sixes")[
    ["Player", "Season", "Sixes", "Runs", "Strike_Rate"]
]
display(top_six_hitters.reset_index(drop=True))
```

```python
qualified_players = df[df["Balls_Faced"] >= 100]
top_strikers = qualified_players.nlargest(10, "Strike_Rate")[
    ["Player", "Season", "Balls_Faced", "Runs", "Strike_Rate"]
]
display(top_strikers.reset_index(drop=True).round(2))
print("Qualification rule: minimum 100 balls faced in the season")
```

```python
runs_distribution = df.groupby("Season")["Runs"].agg(["mean", "median", "max"])
display(runs_distribution.round(2))


phase_totals = df[["Powerplay_Runs", "Middle_Overs_Runs", "Death_Overs_Runs"]].sum()
print("Phase-wise run totals (player-season aggregates):")
display(phase_totals.round(0))
```
