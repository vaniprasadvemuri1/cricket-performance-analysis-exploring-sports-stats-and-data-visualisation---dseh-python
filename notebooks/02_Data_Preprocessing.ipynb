# Notebook 02 — Data Preprocessing


```python
import pandas as pd
import numpy as np
from pathlib import Path

BASE_DIR = Path.cwd()
if BASE_DIR.name == "notebooks":
    BASE_DIR = BASE_DIR.parent

PROCESSED_DIR = BASE_DIR / "data" / "processed"

summary_path = PROCESSED_DIR / "IPL_Batting_Statistics.csv"
innings_path = PROCESSED_DIR / "IPL_Batting_Innings.csv"

df = pd.read_csv(summary_path)
innings_df = pd.read_csv(innings_path)

print("Datasets loaded successfully!")
display(df.head(5))
display(innings_df.head(5))
print("Player-season shape:", df.shape)
print("Individual innings shape:", innings_df.shape)
```

```python
print("Player-season dataset structure:")
df.info()

print("\nStatistical summary:")
display(df.describe())

print("\nMissing values:")
display(df.isnull().sum())

print("\nDuplicate records:", df.duplicated().sum())
display(innings_df.duplicated().sum())

print("\nUnique values per column:")
display(df.nunique())
```

```python
df = df.drop_duplicates(subset=["Season", "Player"]).copy()
innings_df = innings_df.drop_duplicates(
    subset=["Match_ID", "Innings_Number", "Player"]
).copy()

df["Player"] = df["Player"].str.strip()
innings_df["Player"] = innings_df["Player"].str.strip()

df["Season"] = pd.to_numeric(df["Season"], errors="coerce")
innings_df["Season"] = pd.to_numeric(innings_df["Season"], errors="coerce")

df = df.dropna(subset=["Player", "Season"])
innings_df = innings_df.dropna(subset=["Player", "Season"])

df["Season"] = df["Season"].astype(int)
innings_df["Season"] = innings_df["Season"].astype(int)

innings_df["Match_Date"] = pd.to_datetime(
    innings_df["Match_Date"], errors="coerce"
)

innings_df["Is_Super_Over"] = (
    innings_df["Is_Super_Over"]
    .astype(str)
    .str.lower()
    .isin(["true", "1", "yes"])
)

df = df.reset_index(drop=True)
innings_df = innings_df.reset_index(drop=True)

display(df.head(5))
display(innings_df.head(5))
```

```python
df["Batting_Average"] = np.where(
    df["Dismissals"] > 0, df["Runs"] / df["Dismissals"], np.nan
)
df["Strike_Rate"] = np.where(
    df["Balls_Faced"] > 0, df["Runs"] / df["Balls_Faced"] * 100, np.nan
)
df["Boundary_Percentage"] = np.where(
    df["Balls_Faced"] > 0,
    (df["Fours"] + df["Sixes"]) / df["Balls_Faced"] * 100,
    np.nan,
)
df["Dot_Ball_Percentage"] = np.where(
    df["Balls_Faced"] > 0,
    df["Dot_Balls"] / df["Balls_Faced"] * 100,
    np.nan,
)
df["Runs_Per_Match"] = np.where(
    df["Matches_Played"] > 0, df["Runs"] / df["Matches_Played"], np.nan
)
df["Runs_Per_Innings"] = np.where(
    df["Innings"] > 0, df["Runs"] / df["Innings"], np.nan
)

innings_df["Strike_Rate"] = np.where(
    innings_df["Balls_Faced"] > 0,
    innings_df["Runs"] / innings_df["Balls_Faced"] * 100,
    np.nan,
)

display(
    df[[
        "Player", "Season", "Runs", "Batting_Average",
        "Strike_Rate", "Boundary_Percentage"
    ]].head()
)
```

```python
df.to_csv(PROCESSED_DIR / "IPL_Batting_Statistics_Cleaned.csv", index=False)
innings_df.to_csv(PROCESSED_DIR / "IPL_Batting_Innings_Cleaned.csv", index=False)
print("Cleaned datasets saved to", PROCESSED_DIR)
```
